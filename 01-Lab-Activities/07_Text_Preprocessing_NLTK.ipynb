```python
# If needed:
# %pip install nltk scikit-learn

import re
import string
import nltk
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("averaged_perceptron_tagger_eng")

print("NLTK setup complete.")
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from collections import Counter

text = """Machine learning is changing the way we analyze data.
Natural language processing helps computers work with human language.
Text preprocessing is an important step before many NLP models are trained."""

print(text)
```

```python
# Sentence tokenization
sentences = sent_tokenize(text)
display(pd.DataFrame({"Sentence": sentences}))
```

```python
# Word tokenization
tokens = word_tokenize(text)
print(tokens)
print("Number of tokens:", len(tokens))
```

```python
# Lowercase and remove punctuation
lower_tokens = [token.lower() for token in tokens]
clean_tokens = [
    token for token in lower_tokens
    if token.isalpha()
]

print(clean_tokens)
```

```python
# Stopword removal
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    token for token in clean_tokens
    if token not in stop_words
]

print(filtered_tokens)
```

```python
# Stemming
stemmer = PorterStemmer()
stemmed = [stemmer.stem(token) for token in filtered_tokens]

display(pd.DataFrame({
    "Original": filtered_tokens,
    "Stemmed": stemmed
}))
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(token) for token in filtered_tokens]

display(pd.DataFrame({
    "Original": filtered_tokens,
    "Lemmatized": lemmatized
}))
```

```python
# Frequency distribution
freq = Counter(filtered_tokens)
freq_df = pd.DataFrame(freq.most_common(), columns=["Word", "Frequency"])
display(freq_df)
```

```python
# Bigrams and trigrams
from nltk.util import ngrams

bigrams = list(ngrams(filtered_tokens, 2))
trigrams = list(ngrams(filtered_tokens, 3))

print("Bigrams:", bigrams)
print("Trigrams:", trigrams)
```

```python
# Bag-of-Words and TF-IDF
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

documents = [
    "machine learning analyzes data",
    "natural language processing analyzes text",
    "machine learning and natural language processing are useful"
]

bow = CountVectorizer()
X_bow = bow.fit_transform(documents)

bow_df = pd.DataFrame(
    X_bow.toarray(),
    columns=bow.get_feature_names_out()
)
display(bow_df)
```

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(documents)

tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)
display(tfidf_df.round(3))
```

```python
# Small supervised NLP practice
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report, accuracy_score

texts = [
    "I loved this movie and the story was excellent",
    "The film was wonderful and enjoyable",
    "Amazing acting and a great experience",
    "I hated this movie and it was boring",
    "The film was terrible and disappointing",
    "Awful acting and a very boring experience",
    "The movie was good and entertaining",
    "The movie was bad and frustrating",
    "Excellent story and brilliant acting",
    "Poor story and disappointing acting"
]
labels = [1,1,1,0,0,0,1,0,1,0]

X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.30, random_state=42, stratify=labels
)

text_model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("classifier", LogisticRegression(max_iter=1000))
])

text_model.fit(X_train, y_train)
pred = text_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred, zero_division=0))
```
